### **3. Uso de métodos:**
<font size=3>

* Adicione à classe `Animal` um método `adotar()` que altere o atributo `adotado` para `True` e exiba uma mensagem confirmando a adoção.
* Adicione também um método `devolver()` que altere `adotado` para `False` e exiba uma mensagem confirmando a devolução.
* Teste criando um animal, adotando-o e depois devolvendo-o.

````python
#Considere a idade em meses
class Animal:
    def __init__(self, nome, especie, idade):
        self.nome = nome
        self.especie = especie
        self.idade = idade
        self.adotado = False

    def adotar(self):

        if not self.adotado:
            self.adotado = True
            print(f"Parabéns! Você adotou {self.nome}!")
        else:
            print(f"Desculpe, mas o animalzinho {self.nome}' felizmente já encontrou um lar.")

    def devolver(self):

        if self.adotado:
            self.adotado = False
            print(f"Poxa, que pena. Você devolveu {self.nome} para a adoção.")
        else:
            print(f"O animalzinho {self.nome} já está disponível para a adoção.")
````

````python
animal4 = Animal("Farofa", "Canis familiaris", "3 meses")

animal4.adotar()

animal4.devolver()
    
````

### **4. Criando um novo método:**
<font size=3>

* Adicione à classe `Animal` um método `exibir_info()` que mostre:

  ```
  Nome: <nome>
  Espécie: <especie>
  Idade: <idade>
  Adotado: Sim/Não
  ```
* Crie dois animais e teste o método.

````python
#Considere a idade em meses
class Animal:
    def __init__(self, nome, especie, idade):
        self.nome = nome
        self.especie = especie
        self.idade = idade
        self.adotado = False

    def adotar(self):
    
            if not self.adotado:
                self.adotado = True
                print(f"Parabéns! Você adotou {self.nome}!")
            else:
                print(f"Desculpe, mas o animalzinho {self.nome}' felizmente já encontrou um lar.")

    def devolver(self):

        if self.adotado:
            self.adotado = False
            print(f"Poxa, que pena. Você devolveu {self.nome} para a adoção.")
        else:
            print(f"O animalzinho {self.nome} já está disponível para a adoção.")

    def exibir_info(self):

        idade = self.idade

        print(f"Nome: {self.nome}.")
        print(f"Espécie: {self.especie}.")

        if idade < 12:  
            print(f"Idade: {self.idade} mês/meses.")
        else:
            idade_anos = idade/12
            print(f"Idade: {idade_anos} ano/anos.")
            
        if not self.adotado:
            print(f"Adotado: Não.")
        else:
            print(f"Adotado: Sim.")
````

````python
animal5 = Animal("Jeremias", "Diplópode (embuá)", 24)

animal6 = Animal("Bob", "Equus ferus (equino)", 48)

animal5.exibir_info()

animal6.exibir_info()
````

### **5. Atributos de classe:**

  * Pelo nome da classe (`Animal.abrigo`)

````python
#Considere a idade em meses
class Animal:
    abrigo = "Abrigo Esperança"
    def __init__(self, nome, especie, idade):
        self.nome = nome
        self.especie = especie
        self.idade = idade
        self.adotado = False
        
    def adotar(self):
    
            if not self.adotado:
                self.adotado = True
                print(f"Parabéns! Você adotou {self.nome}!")
            else:
                print(f"Desculpe, mas o animalzinho {self.nome}' felizmente já encontrou um lar.")
    
    def devolver(self):

        if self.adotado:
            self.adotado = False
            print(f"Poxa, que pena. Você devolveu {self.nome} para a adoção.")
        else:
            print(f"O animalzinho {self.nome} já está disponível para a adoção.")

    def exibir_info(self):

        idade = self.idade
        
        print(f"Nome: {self.nome}.")
        print(f"Espécie: {self.especie}.")

        if idade < 12:  
            print(f"Idade: {self.idade} mês/meses.")
        else:
            idade_anos = idade/12
            print(f"Idade: {idade_anos} ano/anos.")
            
        if not self.adotado:
            print(f"Adotado: Não.")
        else:
            print(f"Adotado: Sim.")
````

### **6. Lista de animais:**
<font size=3>

* Crie uma lista chamada `lista_animais` contendo pelo menos 5 animais diferentes.

````python
animal1 = Animal("Tixa", "Lagarto", 1)
animal2 = Animal("Clotilda", "Salticidae (come-mosca)", 3)
animal3 = Animal("Sr. Alfredo", "Diplópode (embuá)", 72)
animal4 = Animal("Farofa", "Canis familiaris", 3)
animal5 = Animal("Bob", "Equus ferus (equino)", 48)

lista_animais = [animal1, animal2, animal3, animal4, animal5]
````
