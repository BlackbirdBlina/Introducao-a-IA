## **Enunciado:**
<font size=3>

Implemente um **sistema de diagnóstico** que calcula a probabilidade de um paciente ter uma doença $D$, dado que ele apresenta dois sintomas: Febre ($F$) e Tosse ($T$). Conforme a aula do **_Notebook_ 7**, quando temos múltiplas evidências, assumimos que elas são **condicionalmente independentes** dada a causa,
$$
    P(D \mid F,\, T) = \alpha \cdot P(D) \cdot P(F\mid D) \cdot P(T\mid D) \, ,
$$
onde $\alpha$ é o fator de normalização.

```python
class Sistema_diagnostico:

    def __init__(self, estado_de_crenca):
        
        self.estado_de_crenca = estado_de_crenca

    def decisao(self, febre, tosse):

        doenca = self.estado_de_crenca["P(D)"]
        n_doenca = 1.0 - doenca #P(~D)

        if febre:

            f_dado_D = self.estado_de_crenca["P(F | D)"]
            f_dado_n_D = self.estado_de_crenca["P(F | ~D)"]
        
        else:

            f_dado_D = 1.0 - self.estado_de_crenca["P(F | D)"]
            f_dado_n_D = 1.0 - self.estado_de_crenca["P(F | ~D)"]

        if tosse:
        
            t_dado_D = self.estado_de_crenca["P(T | D)"]
            t_dado_n_D = self.estado_de_crenca["P(T | ~D)"]
        
        else:

            t_dado_D = 1.0 - self.estado_de_crenca["P(T | D)"]
            t_dado_n_D = 1.0 - self.estado_de_crenca["P(T | ~D)"]

        aux1 = doenca * f_dado_D * t_dado_D
        aux2 = n_doenca * f_dado_n_D * t_dado_n_D

        alpha = 1.0 / (aux1 + aux2)

        doenca_dado_F_e_T = alpha * aux1 #P(D = True | F, T)
        n_doenca_dado_F_e_T = alpha * aux2 #P(D = False | F, T)

        return doenca_dado_F_e_T, n_doenca_dado_F_e_T
```

### **Questão 1:**

**Justifique**, em uma **célula _markdown_**, a dependência e independência entre as variáveis.

De acordo com o enunciado, assumimos que Febre ($F$) e Tosse ($T$) são **condicionalmente independentes** dada a Doença ($D$). Com isso, se um paciente está acometido por uma determinada doença $D$, saber que ele tem febre ($F$) não é uma informação a mais em relação à tosse ($T$), o mesmo vale no sentido contrário, pois $P(F,\,T\,|\,D)\,=\,P(F\,|\,D) \cdot P(T\,|\,D)$, seguindo o exemplo dado no [notebook 7 - 4.1 e 4.2](https://colab.research.google.com/drive/1j2mwOH7LGpL9wKnoYR2zwc9H--tAvaag). 

Dessa forma, de acordo com essa expressão: $P(F,\,T\,|\,D)\,=\,P(F\,|\,D) \cdot P(T\,|\,D)$, se existe febre ($F$) e tosse ($T$), ambas são **dependentes** de $D$, já que a probabilidade delas ocorrerem, em conjunto, é baseada ou depende da ocorrência da doença ($D$), ou seja, a probabilidade de $F$ e $T$ ocorrerem é calculada a partir da ocorrência de $D$, em outras palavras, quando $D$ é verdade, caso contrário, as chances desses sintomas ocorrerem devem diminuir consideravelmente.

Isso posto, olhando agora para a variável $D$ (doença), ela está sendo calculada a partir da ocorrência de $F$ (febre) e $T$ (tosse), isto é, quando $(F,\,T)$ é verdade. Dessa forma, $(F,\,T)$ influenciam na probabilidade de $D$ (doença) ocorrer.

Diante disso, podemos entender que, assim como no exemplo dado no [notebook 7 - 4.2](https://colab.research.google.com/drive/1j2mwOH7LGpL9wKnoYR2zwc9H--tAvaag), em geral, $F$ e $T$ são **independentes** entre si, porém quando $D$ é conhecida passam a ser sintomas **dependentes**, pois a probabilidade é calculada quando os dois ocorrem ao mesmo tempo. De modo geral, fazendo uma leitura mais informal dessa expressão, calculamos a probabilidade de uma doença ocorrer quando ela causa febre e tosse ao mesmo tempo.
