````python
class Problem:

    def __init__(self, initial_state, player, actions, result, terminal_test, utility):

        self.initial_state = initial_state
        self.player = player
        self.actions = actions
        self.result = result
        self.terminal_test = terminal_test
        self.utility = utility
````

````python
class Minimax:
    def __init__(self, problem):
        self.problem = problem

    def minimax_decision(self, state):

        best_play = None
        best_value = -np.inf

        for a in self.problem.actions(state):
            value = self.min_value(self.problem.result(state, a))

            if value > best_value:
                best_value = value
                best_play = a

        return best_play

    def max_value(self, state):

        if self.problem.terminal_test(state):
            return self.problem.utility(state)

        v = -np.inf
        for a in self.problem.actions(state):
            v = max(v, self.min_value(self.problem.result(state, a)))

        return v

    def min_value(self, state):

        if self.problem.terminal_test(state):
            return self.problem.utility(state)

        v = np.inf
        for a in self.problem.actions(state):
            v = min(v, self.max_value(self.problem.result(state, a)))

        return v
````
