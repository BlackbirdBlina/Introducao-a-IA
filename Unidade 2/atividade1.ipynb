````python
import numpy as np
````

````python
class Problem:

    def __init__(self, initial_state, player, actions, result, terminal_test, utility):

        self.initial_state = initial_state
        self.player = player
        self.actions = actions
        self.result = result
        self.terminal_test = terminal_test
        self.utility = utility
````

````python
class Minimax:
    def __init__(self, problem):
        self.problem = problem

    def decision(self, state):

        current_player = self.problem.player(state)

        best_play = None
        best_value = -np.inf

        if current_player == "X":
            best_play = None
            best_value = -np.inf

            for a in self.problem.actions(state):
                value = self.min_value(self.problem.result(state, a))

                if value > best_value:
                    best_value = value
                    best_play = a

            return best_play

        else:
            best_play = None
            best_value = np.inf

            for a in self.problem.actions(state):
                value = self.max_value(self.problem.result(state, a))

                if value < best_value:
                    best_value = value
                    best_play = a

            return best_play
    
    def max_value(self, state):

        if self.problem.terminal_test(state):
            return self.problem.utility(state)

        v = -np.inf
        for a in self.problem.actions(state):
            v = max(v, self.min_value(self.problem.result(state, a)))

        return v

    def min_value(self, state):

        if self.problem.terminal_test(state):
            return self.problem.utility(state)

        v = np.inf
        for a in self.problem.actions(state):
            v = min(v, self.max_value(self.problem.result(state, a)))

        return v
````

**Desenvolva** a classe `TicTacToe` como **problema** do Jogo da Velha. Para isso:
- O problema `TicTacToe` irá herdar a classe `Problem`;
- **Implemente** `_player_fn`, `_actions_fn`, `_result_fn`, `_terminal_test_fn`,  `_utility_fn` e `display` seguindo as **dicas** contidas comentadas em cada método.
  

````python
class TicTacToe(Problem):

    def __init__(self):
        initial_board = tuple([tuple(3*[0]) for _ in range(3)])
        initial_state = (initial_board, 'X')

        super().__init__(initial_state = initial_state,
                         player = self._player_fn,
                         actions = self._actions_fn,
                         result = self._result_fn,
                         terminal_test = self._terminal_test_fn,
                         utility = self._utility_fn)

    def _player_fn(self, state):
        """Retorna o jogador."""
        
        board, player = state

        return player


    def _actions_fn(self, state):
        """Retorna as possíveis jogadas do tabuleiro."""

        board, player = state
        
        free_spaces = []
        
        for x in range(3):
            for y in range(3):
                if board[x][y] == 0:
                    free_spaces.append((x, y))

        return free_spaces


    def _result_fn(self, state, action):
        """Retorna o próximo estado dado o estado atual e a ação."""

        board, player = state
        x, y = action

        new_board = [list(row) for row in board]

        if player == "X":
            new_board[x][y] = 1
            next_player = "O"
        else:
            new_board[x][y] = -1
            next_player = "X"

        new_board = tuple(tuple(row) for row in new_board)
        
        return (new_board, next_player)

    def _terminal_test_fn(self, state):
        """Reatorna uma variável booleana para a terminalidade do jogo."""

        board, player = state

        for x in range(3):
            soma = 0

            for y in range(3):
                soma += board[x][y]

            if soma == 3:    
                return True

            elif soma == -3:
                return True

        for x in range(3):
            soma = 0

            for y in range(3):
                soma += board[y][x]

            if soma == 3:    
                return  True

            elif soma == -3:
                return True
            
        soma = board[0][0] + board[1][1] + board[2][2]

        if soma == 3:    
            return True

        elif soma == -3:
            return True

        soma = board[0][2] + board[1][1] + board[2][0]

        if soma == 3:    
            return True

        elif soma == -3:
            return True

        if not self._actions_fn(state):
            return True

        return False

    def _utility_fn(self, state):
        """Retorna o valor correspondente ao estado final do tabuleiro."""

        board, player = state

        for x in range(3):
            soma = 0

            for y in range(3):
                soma += board[x][y]

            if soma == 3:    
                return 1

            elif soma == -3:
                return -1

        for x in range(3):
            soma = 0

            for y in range(3):
                soma += board[y][x]

            if soma == 3:    
                return 1

            elif soma == -3:
                return -1
            
        soma = board[0][0] + board[1][1] + board[2][2]

        if soma == 3:    
            return 1

        elif soma == -3:
            return -1

        soma = board[0][2] + board[1][1] + board[2][0]

        if soma == 3:    
            return 1

        elif soma == -3:
            return -1

        return 0

    def display(self, state):
        """Imprime na tela o tabuleiro"""

        board, player = state

        for x in range(3):
            for y in range(3):

                if board[x][y] == 1:
                    print("X", end = " ")

                elif board[x][y] == -1:
                    print("O", end = " ")

                else:
                    print(".", end = " ")
                    
            print()
````
